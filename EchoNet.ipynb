```python
import cv2
import numpy as np
import os

def video_to_frames(video_path, num_frames=32, resize=(112, 112)):
    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    indices = set(np.linspace(0, total_frames - 1, num_frames, dtype=int))

    frames = []
    i = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        if i in indices:
            frame = cv2.resize(frame, resize)
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            frames.append(frame)
        i += 1
    cap.release()

    while len(frames) < num_frames:
        frames.append(frames[-1] if frames else np.zeros((*resize, 3), dtype=np.uint8))

    return np.array(frames[:num_frames])

processed_dir = '/content/EchoNet_sample/processed'
os.makedirs(processed_dir, exist_ok=True)

sample_video_dir = '/content/EchoNet_sample/Videos'
count = 0
for filename in os.listdir(sample_video_dir):
    if filename.endswith('.avi'):
        video_path = f'{sample_video_dir}/{filename}'
        npy_name = filename.replace('.avi', '.npy')
        npy_path = f'{processed_dir}/{npy_name}'
        if not os.path.exists(npy_path):
            frames = video_to_frames(video_path)
            np.save(npy_path, frames)
            count += 1

print(f"Preprocessed {count} videos")
print("Sample files:", os.listdir(processed_dir)[:5])
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
import pandas as pd
import numpy as np

class EchoNetDataset(Dataset):
    def __init__(self, labels_df, processed_dir):
        self.processed_dir = processed_dir
        # Only keep rows where the .npy file actually exists
        available = set(f.replace('.npy', '') for f in os.listdir(processed_dir))
        self.labels_df = labels_df[labels_df['FileName'].isin(available)].reset_index(drop=True)
        print(f"Dataset ready with {len(self.labels_df)} samples")

    def __len__(self):
        return len(self.labels_df)

    def __getitem__(self, idx):
        row = self.labels_df.iloc[idx]
        filename = row['FileName']
        ef = row['EF']

        frames = np.load(f'{self.processed_dir}/{filename}.npy')
        frames = torch.tensor(frames, dtype=torch.float32).permute(0, 3, 1, 2) / 255.0

        return frames, torch.tensor(ef, dtype=torch.float32)

sample_df = pd.read_csv('/content/EchoNet_sample/labels.csv')
processed_dir = '/content/EchoNet_sample/processed'

dataset = EchoNetDataset(sample_df, processed_dir)
dataloader = DataLoader(dataset, batch_size=4, shuffle=True)

print("Number of batches:", len(dataloader))
```

```python
import torch.nn as nn

class SimpleEFPredictor(nn.Module):
    def __init__(self):
        super().__init__()
        self.cnn = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.AdaptiveAvgPool2d((4, 4))
        )
        self.lstm = nn.LSTM(32 * 4 * 4, 64, batch_first=True)
        self.fc = nn.Linear(64, 1)

    def forward(self, x):
        # x shape: (batch, frames, channels, height, width)
        batch, frames, c, h, w = x.shape
        x = x.view(batch * frames, c, h, w)
        x = self.cnn(x)
        x = x.view(batch, frames, -1)
        x, _ = self.lstm(x)
        x = self.fc(x[:, -1, :])
        return x.squeeze()

device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = SimpleEFPredictor().to(device)
print("Model created on:", device)
print("Total parameters:", sum(p.numel() for p in model.parameters()))
```

```python
import torch.nn as nn

class SimpleEFPredictor(nn.Module):
    def __init__(self):
        super().__init__()
        self.cnn = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.AdaptiveAvgPool2d((4, 4))
        )
        self.lstm = nn.LSTM(32 * 4 * 4, 64, batch_first=True)
        self.fc = nn.Linear(64, 1)

    def forward(self, x):
        batch, frames, c, h, w = x.shape
        x = x.reshape(batch * frames, c, h, w)
        x = self.cnn(x)
        x = x.reshape(batch, frames, -1)
        x, _ = self.lstm(x)
        x = self.fc(x[:, -1, :])
        return x.squeeze()

device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = SimpleEFPredictor().to(device)
print("Model re-created on:", device)
```

```python
import torch.optim as optim
import os

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

os.makedirs('/content/drive/MyDrive/checkpoints', exist_ok=True)

num_epochs = 10
for epoch in range(num_epochs):
    model.train()
    total_loss = 0
    for frames, ef in dataloader:
        frames, ef = frames.to(device), ef.to(device)

        optimizer.zero_grad()
        predictions = model(frames)
        loss = criterion(predictions, ef)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(dataloader)
    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {avg_loss:.4f}")

    torch.save(model.state_dict(),
               f'/content/drive/MyDrive/checkpoints/ef_model_epoch{epoch+1}.pt')

print("Training complete")
```

```python
import torchvision.models.video as video_models
import torch.nn as nn

class EFPredictor3D(nn.Module):
    def __init__(self):
        super().__init__()
        # Load pretrained 3D ResNet (trained on real video data)
        self.backbone = video_models.r3d_18(weights=video_models.R3D_18_Weights.DEFAULT)
        # Replace final classification layer with a regression head
        self.backbone.fc = nn.Linear(self.backbone.fc.in_features, 1)

    def forward(self, x):
        # x shape needed: (batch, channels, frames, height, width)
        return self.backbone(x).squeeze()

device = 'cuda' if torch.cuda.is_available() else 'cpu'
model_3d = EFPredictor3D().to(device)
print("3D-CNN model created on:", device)
print("Total parameters:", sum(p.numel() for p in model_3d.parameters()))
```

```python
class EchoNetDataset3D(Dataset):
    def __init__(self, labels_df, processed_dir):
        self.processed_dir = processed_dir
        available = set(f.replace('.npy', '') for f in os.listdir(processed_dir))
        self.labels_df = labels_df[labels_df['FileName'].isin(available)].reset_index(drop=True)
        print(f"Dataset ready with {len(self.labels_df)} samples")

    def __len__(self):
        return len(self.labels_df)

    def __getitem__(self, idx):
        row = self.labels_df.iloc[idx]
        filename = row['FileName']
        ef = row['EF'] / 100.0

        frames = np.load(f'{self.processed_dir}/{filename}.npy')
        frames = torch.tensor(frames, dtype=torch.float32) / 255.0
        frames = frames.permute(3, 0, 1, 2)  # (channels, frames, height, width)

        return frames, torch.tensor(ef, dtype=torch.float32)

sample_df_1000 = pd.read_csv('/content/EchoNet_sample1000/labels.csv')
processed_dir_1000 = '/content/EchoNet_sample1000/processed'

train_df_1000 = sample_df_1000[sample_df_1000['Split'] == 'TRAIN']
val_df_1000 = sample_df_1000[sample_df_1000['Split'] == 'VAL']

print(f"Train: {len(train_df_1000)}, Val: {len(val_df_1000)}")

train_dataset_3d = EchoNetDataset3D(train_df_1000, processed_dir_1000)
val_dataset_3d = EchoNetDataset3D(val_df_1000, processed_dir_1000)

train_loader_3d = DataLoader(train_dataset_3d, batch_size=2, shuffle=True)
val_loader_3d = DataLoader(val_dataset_3d, batch_size=2, shuffle=False)
```

```python
import torchvision.models.video as video_models
import torch.nn as nn
import torch.optim as optim
import os

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print("Device:", device)

class EFPredictor3D(nn.Module):
    def __init__(self):
        super().__init__()
        self.backbone = video_models.r3d_18(weights=video_models.R3D_18_Weights.DEFAULT)
        self.backbone.fc = nn.Linear(self.backbone.fc.in_features, 1)

    def forward(self, x):
        return self.backbone(x).squeeze()

model_3d = EFPredictor3D().to(device)
print("Model created")
```

```python
import time

checkpoint_path = '/content/drive/MyDrive/checkpoints/full_training_checkpoint.pt'
criterion = nn.MSELoss()
optimizer_3d = optim.Adam(model_3d.parameters(), lr=1e-4)

start_epoch = 0
start_batch = 0

# Resume if a checkpoint already exists
if os.path.exists(checkpoint_path):
    ckpt = torch.load(checkpoint_path)
    model_3d.load_state_dict(ckpt['model_state_dict'])
    optimizer_3d.load_state_dict(ckpt['optimizer_state_dict'])
    start_epoch = ckpt['epoch']
    print(f"Resumed from epoch {start_epoch}")
else:
    print("No checkpoint found, starting fresh")

num_epochs = 10  # full dataset needs fewer epochs than small samples

for epoch in range(start_epoch, num_epochs):
    model_3d.train()
    epoch_start = time.time()
    running_loss = 0

    for batch_idx, (frames, ef) in enumerate(train_loader_full):
        frames, ef = frames.to(device), ef.to(device)

        optimizer_3d.zero_grad()
        predictions = model_3d(frames)
        loss = criterion(predictions, ef)
        loss.backward()
        optimizer_3d.step()

        running_loss += loss.item()

        if batch_idx % 200 == 0:
            torch.save({
                'epoch': epoch,
                'model_state_dict': model_3d.state_dict(),
                'optimizer_state_dict': optimizer_3d.state_dict(),
            }, checkpoint_path)
            elapsed = (time.time() - epoch_start) / 60
            print(f"Epoch {epoch+1}, Batch {batch_idx}/{len(train_loader_full)}, "
                  f"Loss: {running_loss/(batch_idx+1):.5f}, Time: {elapsed:.1f} min")

    # Validation at end of epoch
    model_3d.eval()
    val_loss = 0
    with torch.no_grad():
        for frames, ef in val_loader_full:
            frames, ef = frames.to(device), ef.to(device)
            predictions = model_3d(frames)
            val_loss += criterion(predictions, ef).item()

    avg_val_loss = val_loss / len(val_loader_full)
    print(f"=== Epoch {epoch+1} complete. Val Loss: {avg_val_loss:.5f} ===")

    torch.save({
        'epoch': epoch + 1,
        'model_state_dict': model_3d.state_dict(),
        'optimizer_state_dict': optimizer_3d.state_dict(),
    }, checkpoint_path)

print("Full training complete")
```
